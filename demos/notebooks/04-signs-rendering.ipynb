# Render a complete signs-not-directions trace

**Evidence state:** Defined

## What this demonstrates

Generate the canonical VPM bundle, JSON decision trace, full policy surface, annotated FIRE decision, and animated replay from one bounded policy artifact.

## Why it matters

A policy artifact becomes easier to review when the state, addressed VPM cell, candidate values, selected action, proof coordinates, and complete trace are visible together.


````python
from pathlib import Path
import os
import sys

ROOT = Path.cwd().resolve()
while not (ROOT / "VERSION").is_file():
    if ROOT.parent == ROOT:
        raise RuntimeError("ZeroModel repository root not found")
    ROOT = ROOT.parent
os.chdir(ROOT)
for path in (ROOT, ROOT / "examples"):
    if str(path) not in sys.path:
        sys.path.insert(0, str(path))
print(f"repository root: {ROOT}")
````

## Source and package mapping

- `examples/render_signs_demo.py`
- `examples/arcade_shooter_policy.py`
- `zeromodel`
- `zeromodel-video`


````python
import json
from IPython.display import Image, display
from examples.render_signs_demo import main as render_signs

output = ROOT / "build" / "demo-notebooks" / "signs-rendering"
render_signs(
    [
        "--output-dir",
        str(output),
        "--dpi",
        "80",
        "--frame-duration-ms",
        "300",
    ]
)
print(
    json.dumps(
        {path.name: path.stat().st_size for path in sorted(output.iterdir())}, indent=2
    )
)
````

````python
display(Image(filename=str(output / "zero_policy_vpm.png"), width=260))
display(Image(filename=str(output / "zero_money_shot.png"), width=900))
display(Image(filename=str(output / "zero_replay.gif"), width=900))
````

## Application

```text
policy artifact -> addressed decisions -> rendered proof frames -> reviewable replay
```

The same output family can support documentation, policy debugging, review, and demonstrations without changing policy semantics.

## Boundaries and limitations

The rendering proves that the bounded trace is reproducible and source-addressable. Raster details can vary with font and plotting-library versions, and the fixture does not establish open-world policy quality.

## Reproduction record

The builder records the executed notebook, ZeroModel and dependency versions, Git revision, platform, command, and generated HTML under `docs/results/demos/signs-rendering/`.


````python
summary = json.loads((output / "zero_policy_results.json").read_text(encoding="utf-8"))
print(
    json.dumps(
        {
            "demo_id": "signs-rendering",
            "artifact_id": summary["artifact_id"],
            "policy_cleared": summary["policy_cleared"],
            "policy_steps": summary["policy_steps"],
        },
        indent=2,
    )
)
````
