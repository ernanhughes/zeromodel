# Localize and repair a policy counterexample

**Evidence state:** Measured

## What this demonstrates

Compile a Q-bearing policy with criticality evidence, verify finite properties, inject one unsafe FIRE preference, localize its counterexample, repair the policy, and verify it again.

## Why it matters

Verification becomes operational when a failed property produces an identity-bearing artifact and a concrete source location rather than only a pass/fail label.


````python
from pathlib import Path
import os
import sys

ROOT = Path.cwd().resolve()
while not (ROOT / "VERSION").is_file():
    if ROOT.parent == ROOT:
        raise RuntimeError("ZeroModel repository root not found")
    ROOT = ROOT.parent
os.chdir(ROOT)
for path in (ROOT, ROOT / "examples"):
    if str(path) not in sys.path:
        sys.path.insert(0, str(path))
print(f"repository root: {ROOT}")
````

## Source and package mapping

- `examples/criticality_verification.py`
- `zeromodel`
- `zeromodel-analysis`


````python
from importlib.metadata import version
import json
from IPython.display import Image, display
from examples.criticality_verification import run_demo

output = ROOT / "build" / "demo-notebooks" / "criticality-verification"
result = run_demo(output)
public_result = {
    "zeromodel_version": version("zeromodel"),
    "original_passed": result["original_passed"],
    "unsafe_passed": result["unsafe_passed"],
    "repaired_passed": result["repaired_passed"],
    "counterexample_count": len(result["counterexamples"]),
    "counterexamples": result["counterexamples"],
}
print(json.dumps(public_result, indent=2, sort_keys=True))
````

````python
display(Image(filename=str(output / "policy_original.png"), width=260))
display(Image(filename=str(output / "verification_failed.png"), width=260))
display(Image(filename=str(output / "verification_repaired.png"), width=260))
````

## Application

```text
policy + declared property -> exhaustive finite checker -> localized violation
                              -> repair -> linked verification artifact
```

This pattern applies to bounded routing, safety gates, state machines, and other finite policies whose acceptable behavior can be stated explicitly.

## Boundaries and limitations

The checker exhausts this finite arcade fixture and these two declared properties. It does not prove properties that were not declared, nor does it generalize the fixture to an open environment.

## Reproduction record

The generated result includes original, unsafe, and repaired policy identities; linked verification artifacts; counterexamples; and the build environment recorded by the demo pipeline.
