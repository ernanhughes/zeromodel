# Recover planted matrix structure

**Evidence state:** Measured

## What this demonstrates

Run the existing Bertin-inspired example over a shuffled three-block matrix, calibrate the selected ordering against a permutation null, and materialize both a report artifact and a discovered VPM view.

## Why it matters

A discovered arrangement should carry its source data, declared method, statistical calibration, identities, and explicit order rather than appearing as an unexplained heatmap.


````python
from pathlib import Path
import os
import sys

ROOT = Path.cwd().resolve()
while not (ROOT / "VERSION").is_file():
    if ROOT.parent == ROOT:
        raise RuntimeError("ZeroModel repository root not found")
    ROOT = ROOT.parent
os.chdir(ROOT)
for path in (ROOT, ROOT / "examples"):
    if str(path) not in sys.path:
        sys.path.insert(0, str(path))
print(f"repository root: {ROOT}")
````

## Source and package mapping

- `examples/bertin_pattern_detection.py`
- `zeromodel`
- `zeromodel-analysis`


````python
import json
import subprocess
import sys
from IPython.display import Image, display

output = ROOT / "build" / "demo-notebooks" / "bertin-pattern-detection"
completed = subprocess.run(
    [
        sys.executable,
        "examples/bertin_pattern_detection.py",
        "--output-dir", str(output),
        "--null-samples", "99",
        "--seed", "11",
        "--alpha", "0.05",
    ],
    cwd=ROOT,
    check=True,
    capture_output=True,
    text=True,
)
result = json.loads(completed.stdout)
print(json.dumps(result, indent=2, sort_keys=True))
display(Image(filename=str(output / "pattern-view.png"), width=420))
````

## Application

```text
unchanged score table -> deterministic seriation -> corrected null calibration
                      -> report artifact + explicit-order view artifact
```

This can support exploratory matrix inspection while preserving the distinction between detected geometry and domain interpretation.

## Boundaries and limitations

The matrix contains deliberately planted blocks. The detector recovers geometry in this bounded fixture but does not assign semantic meaning, establish causality, or prove usefulness on arbitrary datasets.
